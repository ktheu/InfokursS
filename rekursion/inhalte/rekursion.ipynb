## Rekursion

[Video](https://youtu.be/A3H1Gbx0Q0E) - [Video2](https://youtu.be/vdDzbicAKQA) - [AufgabenVideo](https://youtu.be/hFQ4JdnN2kE)

Eine Funktion darf sich im Rumpf selbst wieder aufrufen. Man nennt das
einen **rekursiven Aufruf**. Jede Rekursion braucht eine Bremse, damit es
nicht zu unendlich vielen Aufrufen kommt.

Jede rekursive Funktion besteht aus zwei Teilen:

- **Rekursionsbasis** (Basisfall, Abbruchbedingung): ein einfacher Fall, bei dem das Ergebnis sofort feststeht, ohne rekursiven Aufruf.
- **Rekursionsschritt** (rekursiver Fall): Die Funktion ruft sich selbst mit einem *kleineren* Problem auf. Jeder Aufruf muss der Rekursionsbasis ein Stück näher kommen.

#### Beispiel 1: Fakultät
Iterative Definition der Fakultät:  
$n! =\begin{cases}
 1   &  n = 0 \\
 1 \cdot 2 \cdot 3 \cdot ... \cdot n & n > 0 
\end{cases} $

Rekursive Definition der Fakultät: 

$n! =\begin{cases}
 1   & n = 0 \\ 
 n \cdot (n-1)!  & n > 0 
\end{cases} $

 Rekursive Berechnung von $5!$ :

<img src='fakultaet.png' width='400'>

```python
def fakultaet_iterativ(n):
    if n == 0: return 1
    tmp = 1
    for k in range(1,n+1):
        tmp *= k
    return tmp

print(fakultaet_iterativ(5))
```

```python
def fakultaet(n):
    if n == 0: return 1
    return n * fakultaet(n-1)

print(fakultaet(5))
```

Was passiert bei den Aufrufen? Wir geben bei jedem Aufruf und bei jedem `return` eine Zeile aus.
Der Parameter `tiefe` sorgt nur für die Einrückung.

```python
def fakultaet_ablauf(n, tiefe=0):
    print('    ' * tiefe + f'fakultaet({n})')
    if n == 0:
        ergebnis = 1
    else:
        ergebnis = n * fakultaet_ablauf(n-1, tiefe+1)
    print('    ' * tiefe + f'fakultaet({n}) gibt {ergebnis} zurück')
    return ergebnis

fakultaet_ablauf(4)
```

Jeder Aufruf muss warten, bis der Aufruf darunter sein Ergebnis zurückgibt. Erst wenn der Basisfall erreicht ist,
werden die Ergebnisse von unten nach oben zurückgegeben und verrechnet.

Ohne Basisfall ruft sich die Funktion immer weiter selbst auf. Python bricht nach einer festen Höchstzahl geschachtelter
Aufrufe mit einem `RecursionError` ab (normalerweise 1000, in Jupyter 3000).

```python
def fakultaet_ohne_basisfall(n):
    return n * fakultaet_ohne_basisfall(n-1)

# try/except fängt den Fehler ab, damit das Notebook weiterläuft
try:
    fakultaet_ohne_basisfall(5)
except RecursionError as fehler:
    print('RecursionError:', fehler)
```

Dasselbe passiert, wenn es zwar einen Basisfall gibt, er aber nie erreicht wird, z.B. bei `fakultaet(-1)`:
-1, -2, -3, ... wird nie 0. Sicherer ist deshalb oft `if n <= 0: return 1`.

#### Beispiel 2: Zweierpotenz

Iterative Definition der Zweierpotenz:

$2^n =\begin{cases}
 1   &  n = 0 \\
 2 \cdot 2 \cdot 2 \cdot ... \cdot 2 \text{  (n-mal)} & n > 0 
\end{cases} $

```python
def zweihoch_iterativ(n):
    if n == 0: return 1
    tmp = 1
    for k in range(n):
        tmp *= 2
    return tmp

zweihoch_iterativ(10)
```

Rekursive Definition der Zweierpotenz:

$2^n =\begin{cases}
 1,   & n = 0 \\
 2 \cdot 2^{n-1}  & n > 0 
\end{cases} $

```python
def zweihoch(n):
    if n == 0: return 1
    return 2 * zweihoch(n-1)

zweihoch(10)
```

#### Beispiel 3: 

Die rekursive Funktion *dreheUm(s)* gibt den String s in umgedrehter Reihenfolge zurück

<img src='servus1.png'  width='700'>

Beim Aufstellen der Rekursionsbeziehung dürfen wir auf der rechten Seite die Funktion, die wir definieren wollen, mit einer kleineren Problemstellung wiederverwenden (Rekurrenzrelation, recurrence = erneutes Auftreten).

<img src='rek1.png' width='700'>

Von der Rekursionsbeziehung zum fertigen Programm ist es meist nur ein kleiner Schritt

```python
def dreheUm(s):
    if len(s) == 0: return ''
    return s[-1]+dreheUm(s[:-1])

dreheUm('Servus')
```

Wir könnten die Problemgröße auch verkleinern, wenn wir den ersten Buchstaben weglassen:

<img src='servus2.png' width='700'>

Das führt zur Rekursionsbeziehung:

<img src='rek2.png' width='700'>

```python
def dreheUm(s):
    if len(s) == 0: return ''
    return dreheUm(s[1:]) + s[0]

dreheUm('Servus')
```

#### Das übliche Muster bei Strings und Listen

Wir bearbeiten das erste Element `s[0]` und überlassen den Rest `s[1:]` dem rekursiven Aufruf.
Der Basisfall ist der leere String bzw. die leere Liste.

```python
def ohne_leerzeichen(s):
    if len(s) == 0: return ''
    if s[0] == ' ':
        return ohne_leerzeichen(s[1:])
    return s[0] + ohne_leerzeichen(s[1:])

print(ohne_leerzeichen('Das ist ein Test'))
```

```python
def summe_liste(a):
    if len(a) == 0: return 0
    return a[0] + summe_liste(a[1:])

print(summe_liste([3, 5, 2, 7]))
```

#### Beispiel 4: 

Die rekursive Funktion *summe(n)* gibt die Summe der natürlichen Zahlen von 1 bis n (einschließlich) zurück.

<img src='rek3.png' width='700'>

```python
def summe(n):
    if n == 1: return 1
    return summe(n-1) + n

summe(100)
```

#### Beispiel 6: Binäre Suche

<img src='rek6.png' width='852'>

Die rekursive Funktion wird manchmal mit mit mehr Parametern definiert als wir für das Ausgangsproblem benötigen, denn wir müssen in der Parameterliste alles vorsehen, was wir für die Anwendung auf kleinere Problemgrößen benötigen.

```python
def binaereSuche(a,x,L,R):
    '''
    a: sortierte Liste mit Zahlen
    x: Zahl
    L,R: ints zwischen 0 und len(a)-1
    returns: Index von x in a, falls x in a
             -1              , falls x nicht in a
    '''
    if L > R: return -1
    mid = L + (R - L) // 2
    if a[mid] == x: return mid
    if x < a[mid]:
        return binaereSuche(a,x,L,mid-1)
    else:
        return binaereSuche(a,x,mid+1,R)

a = [2, 4, 6, 12, 22, 42, 49]
binaereSuche(a,42,0,len(a)-1)
```

#### Türme von Hanoi

<img src="rekursion_01.png" width="300"/>

[Video](https://www.youtube.com/watch?v=w9LgLiW9YHU)

<img src="hanoi_rek.png" width="800"/>

Die Rekursionsidee:

<img src="hanoi_idee.png" width="701"/>

Die Rekursionsbeziehung:

<img src="hanoi_rek2.png" width="800"/>

```python
def hanoi(n, start, ziel, zwischen):
    '''
    n: Anzahl der Scheiben
    start, ziel, zwischen: Strings, die 3 Stapel bezeichnen
    '''
    if n == 0: return
    hanoi(n-1,start,zwischen,ziel)
    print("Scheibe",n,"von",start,"nach",ziel)
    hanoi(n-1,zwischen,ziel,start)
```

```python
hanoi(4,'A','C','B')
```

#### Divide and Conquer

Viele rekursive Algorithmen folgen dem Prinzip **Divide and Conquer** (teile und herrsche):

1. **Teilen:** Wir zerlegen das Problem in kleinere Teilprobleme derselben Art.
2. **Lösen:** Wir lösen die Teilprobleme rekursiv. Sehr kleine Probleme lösen wir direkt.
3. **Zusammenfügen:** Aus den Teillösungen bilden wir die Gesamtlösung.

Beispiel: Wir suchen das Maximum einer Liste. Wir teilen die Liste in zwei Hälften, suchen in jeder Hälfte das Maximum und nehmen das größere der beiden.

```python
def maximum(a):
    if len(a) == 1: return a[0]        # Rekursionsbasis
    mitte = len(a) // 2
    links = maximum(a[:mitte])         # Teilprobleme lösen
    rechts = maximum(a[mitte:])
    if links > rechts:                 # zusammenfügen
        return links
    return rechts

print(maximum([4, 9, 2, 7, 5]))
```

Die **binäre Suche** ist ein Sonderfall: Wir teilen den Suchbereich und suchen nur in einem Teil weiter, das Zusammenfügen entfällt.
Auch bei den **Türmen von Hanoi** zerlegen wir das Problem in kleinere Probleme derselben Art. Ein wichtiges Beispiel ist das Sortierverfahren **MergeSort**, das wir noch kennenlernen.

#### Zusammenfassung

| Begriff | Bedeutung |
|---|---|
| rekursiver Aufruf | Die Funktion ruft sich selbst auf. |
| Rekursionsbasis | Fall ohne rekursiven Aufruf, beendet die Rekursion |
| Rekursionsschritt | Aufruf mit einem kleineren Problem, das der Rekursionsbasis näher kommt |
| Rekursionsbeziehung | Beschreibt das Ergebnis mithilfe der Funktion für ein kleineres Problem |
| Rekursionsbaum | Zeigt alle Aufrufe. |
| `RecursionError` | zu viele geschachtelte Aufrufe, meist fehlt der Basisfall oder er wird nie erreicht |
| Divide and Conquer | Problem teilen, Teile rekursiv lösen, Teillösungen zusammenfügen |
