### Variablen und Zuweisungen 

Zunächst betrachten wir zwei Arten von Zahlen:

| Datentyp | In Python | Beispiele |
|:--|:--|:--|
| **Ganze Zahl** | `int` | `8`, `-3`, `0` |
| **Dezimalzahl** (Gleitkommazahl, *floating point number*) | `float` | `3.2`, `-0.5`, `2.0` |

**Dezimalzahlen werden mit einem Punkt geschrieben**, nicht mit einem Komma.


Mit einer **Zuweisung** ordnen wir einer Variablen einen Wert zu. Was auf der rechten Seite des Gleichheitszeichens steht, wird der Variablen auf der linken Seite zugewiesen.

```python
k = 8
x = 3.2
```

Mit der Funktion `print` geben wir Werte aus. `print` kann – mit Komma getrennt – beliebig viele Werte ausgeben. In der Ausgabe stehen sie durch ein Leerzeichen getrennt.

```python
print(k)
print(x)
```

```python
print(k,x,10)
```

Durch eine **erneute Zuweisung** verändern wir den Wert einer Variablen. Der alte Wert ist danach verloren.

```python
x = 5
print(x)
```

```python
x = x + 1
print(x)
```

Für besonders häufige Zuweisungen gibt es **Kurzformen**:

```python
x += 1    # entspricht  x = x + 1
x -= 1    # entspricht  x = x - 1
x *= 2    # entspricht  x = x * 2   … usw.
```

### Variablennamen

**Regeln für Variablennamen:**

- Variablennamen sind **case-sensitiv**: `anzahl` und `Anzahl` sind zwei verschiedene Variablen.
- Das erste Zeichen ist ein **Buchstabe oder ein Unterstrich**, keine Ziffer.
- Wir beschränken uns freiwillig auf den **Unterstrich, die Buchstaben des englischen Alphabets und die Ziffern**. Leerzeichen und Zeichen wie `-` sind nicht erlaubt.
- Schlüsselwörter der Sprache wie `if`, `for` oder `while` sind als Namen verboten.

Wenn es der Lesbarkeit nicht schadet, erlauben wir uns kurze Namen wie `x` oder `k`. Ansonsten nehmen wir **sprechende Namen**. **Variablennamen beginnen in der Regel mit einem kleinen Buchstaben.** Zusammengesetzte Namen schreiben wir entweder in *camelCase* oder mit Unterstrich verbunden (*snake_case*), aber nicht gemischt:
Wenn es der Lesbarkeit des Programms nicht schadet, erlauben wir uns kurze Variablennamen.

```python
zaehl = 1
radius = 2.2
anzahlTreffer = 5            # camelCase
entfernungZurWand = 3.2      # camelCase
anzahl_treffer = 5           # snake_case
entfernung_zur_wand = 3.2    # snake_case
```

### Arithmetische Operatoren

Grundrechenarten, Exponentation, Modulo-Operator, ganzzahlige Division

```python
x1 = 11 + 3
x2 = 11 - 3
x3 = 11 * 3
x4 = 11 / 3
x5 = 11 ** 3           # 11 hoch 3
x6 = 11 % 3            # 11 modulo 3 = Rest bei der Division 11 / 3
x7 = 11 // 3           # ganzzahlige Division
print(x1, x2, x3, x4, x5, x6, x7)
```

**`//` und `%` gehören zusammen:** 11 geteilt durch 3 ergibt 3 Rest 2, denn 11 = 3 · 3 + 2.

**Die Division `/` liefert immer einen `float`**, auch wenn sie aufgeht: `6 / 2` ergibt `3.0`.

**Auswertungsreihenfolge:** Exponentation vor Punktrechnung vor Strichrechnung. Im Zweifelsfall setzen wir Klammern.

### Strings

**Strings** sind Zeichenketten. Sie werden in einfache `'…'` oder doppelte `"…"` Hochkommata eingeschlossen. Beide Schreibweisen sind gleichwertig.

```python
a = 'Hallo Welt'
print(a)
```

```python
b = "Guten Morgen"
print(b)
```

Die **Länge** eines Strings ist die Anzahl seiner Zeichen. Wir erhalten sie mit `len()`. Leerzeichen zählen mit.

```python
s = 'Habicht'
print(len(s))
```

Mit dem Zeichen `+` können wir Strings aneinanderhängen (**konkatenieren**). Leerzeichen werden dabei nicht automatisch eingefügt, wir müssen sie selbst ergänzen.

```python
s1 = 'Hallo'
s2 = 'du'
s3 = 'Welt'
s4 = s1 + ' ' + s2 + ' ' + s3
print(s4)
```

**Vorsicht:** `'3' + '4'` ergibt den String `'34'`, denn Strings werden verkettet, nicht addiert. Einen String und eine Zahl kann man nicht mit `+` verbinden: `'Alter: ' + 15` führt zu einem *TypeError*.

##### Indexing

Jedes Zeichen eines Strings hat **zwei Indizes**: von vorne gezählt positiv, beginnend bei `0`, und von hinten gezählt negativ, beginnend bei `-1`.

<img src='./habicht.png' width='300'>

Mit **Indexing** greifen wir auf ein einzelnes Zeichen zu:

```python
s = 'Habicht'
c1 = s[0]          # erstes Zeichen
c2 = s[1]          # zweites Zeichen
c3 = s[len(s)-1]   # letztes Zeichen
c4 = s[-1]         # letztes Zeichen
c5 = s[-2]         # vorletztes Zeichen
print(c1, c2, c3, c4, c5)
```

Der größte positive Index ist `len(s) - 1`. Ein Zugriff wie `s[7]` führt zu einem *IndexError*.

### Input

Mit der Funktion `input()` lesen wir Eingaben des Benutzers von der Konsole ein. Der Text in den Klammern wird als Aufforderung angezeigt. Das Programm wartet, bis der Benutzer die Eingabe mit Enter abschließt.

```python
s = input('Wie ist dein Vorname?')
print('Hallo',s)
```

**`input()` liefert immer einen String**, auch wenn Ziffern eingegeben werden. Wollen wir eine Eingabe als Zahl behandeln, wandeln wir den String mit `int()` oder `float()` in eine Zahl um.

```python
x = int(input('Bitte eine ganze Zahl eingeben:'))
print('Das Doppelte von',x,'ist',2*x)
```

### f-String

Für die Ausgabe von Text und Variablen nutzen wir einen **f-String**. Er beginnt mit einem `f` vor dem Hochkomma. Variablen oder ganze Ausdrücke schreiben wir in **geschweiften Klammern** direkt in den String. Eine Umwandlung mit `str()` ist nicht nötig.

```python
x = int(input('Bitte eine ganze Zahl eingeben:'))
print(f'Das Doppelte von {x} ist {2*x}.')
```

Bei Bedarf fügen wir nach einem Doppelpunkt eine **Formatierung** hinzu. Mit `:.2f` wird eine Dezimalzahl auf zwei Nachkommastellen gerundet:

```python
# Runden einer Dezimalzahl auf eine bestimmte Anzahl von Nachkommastellen
pi = 3.1415926535
print(f"Pi auf zwei Nachkommastellen: {pi:.2f}")
```

Die Zahl vor dem f gibt die Anzahl der Nachkommastellen an: {pi:.4f} ergibt 3.1416. Der Wert der Variablen selbst bleibt dabei unverändert, gerundet wird nur die Ausgabe.
