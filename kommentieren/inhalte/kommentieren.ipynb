### Code kommentieren

Programme werden viel öfter gelesen als geschrieben: von anderen, die mit unserem Code arbeiten, und von uns selbst, wenn wir nach
ein paar Monaten nicht mehr wissen, was wir uns gedacht haben. Damit Code verständlich bleibt, gibt es drei Mittel:

- **Gute Namen** für Variablen, Funktionen und Klassen
- **Kommentare** für das, was man dem Code nicht ansieht
- **Docstrings**, die beschreiben, wie eine Funktion oder Klasse benutzt wird

Für Python gibt es dazu zwei verbreitete Standards: **PEP 8** ist der Style Guide für Python-Code, **PEP 257** beschreibt
die Konventionen für Docstrings. Eine PEP (*Python Enhancement Proposal*) ist ein Dokument, das einen Vorschlag für Python beschreibt.
Unsere Entwicklungsumgebung hilft uns dabei, diese Standards einzuhalten.

#### Kommentare mit `#`

Python ignoriert alles, was hinter einem `#` steht. Nach PEP 8 schreiben wir Kommentare so:

- Ein **Blockkommentar** steht in einer eigenen Zeile über dem Code, auf den er sich bezieht, und ist genauso eingerückt.
- Ein **Zeilenkommentar** steht hinter dem Code, mit mindestens zwei Leerzeichen Abstand.
- Hinter dem `#` steht ein Leerzeichen.

```python
def ist_schaltjahr(j):
    # Jahrhundertjahre sind nur Schaltjahre, wenn sie durch 400 teilbar sind.
    if j % 100 == 0:
        return j % 400 == 0
    return j % 4 == 0

print(ist_schaltjahr(1900), ist_schaltjahr(2000), ist_schaltjahr(2024))  # False True True
```

#### Was gehört in einen Kommentar?

Ein guter Kommentar erklärt, **warum** etwas so gemacht wird. **Was** gemacht wird, sehen wir im Code. Dieser Kommentar hilft niemandem:

    i = i + 1      # erhöhe i um 1

Der folgende Kommentar dagegen erklärt eine Entscheidung, die man dem Code nicht ansieht:

```python
a = [3, -1, -4, 5]
# von hinten nach vorne, weil pop die Indizes der folgenden Elemente verschiebt
for i in range(len(a)-1, -1, -1):
    if a[i] < 0:
        a.pop(i)
print(a)
```

Sinnvoll sind Kommentare für

- den Grund einer Entscheidung,
- die Idee eines Algorithmus in einem Satz,
- Voraussetzungen, die der Code nicht selbst prüft (`# a muss sortiert sein`),
- Quellen, z. B. die Internetseite, von der eine Formel stammt.

Vermeiden sollten wir

- Kommentare, die nur den Code wiederholen,
- **veraltete Kommentare**: Ein Kommentar, der nicht mehr zum Code passt, ist schlimmer als gar keiner, deshalb ändern wir ihn immer mit,
- auskommentierten Code, den niemand mehr braucht. Den löschen wir.

#### Gute Namen sparen Kommentare

Oft ist ein Kommentar nur nötig, weil ein Name schlecht gewählt ist. Mit guten Namen braucht dasselbe Programm keinen Kommentar.
Feste Zahlen im Code bekommen als **Konstante** einen Namen.

```python
# schlecht: Der Kommentar muss erklären, was die Namen nicht sagen.
def f(t):
    return t * 86400   # t Tage in Sekunden

# besser
SEKUNDEN_PRO_TAG = 86400

def tage_in_sekunden(tage):
    return tage * SEKUNDEN_PRO_TAG

print(f(2), tage_in_sekunden(2))
```

PEP 8 legt fest, wie Namen aussehen:

| Was | Schreibweise | Beispiel |
|---|---|---|
| Variable, Funktion, Methode | Kleinbuchstaben, Wörter mit `_` getrennt | `anzahl_schueler`, `ist_prim` |
| Klasse | jedes Wort mit großem Anfangsbuchstaben | `BankKonto` |
| Konstante | Großbuchstaben, Wörter mit `_` getrennt | `MAX_PUNKTE` |

Kurze Namen wie `i`, `j` oder `x` sind in Ordnung, wenn eine Variable nur in einer kleinen Schleife gebraucht wird.

#### Docstrings

Der **Docstring** ist ein String direkt unter dem Kopf einer Funktion. Er beschreibt, **wie die Funktion benutzt wird**: welche
Argumente sie erwartet und was sie zurückgibt. Mit `help` zeigen wir ihn an. Wer die Funktion benutzt, muss ihren Code nicht lesen.

Wir schreiben Docstrings wie im Abschnitt über Funktionen. Beispielaufrufe machen die Beschreibung verständlich.

```python
def zeichen(k, s='*'):
    '''
    k: positive ganze Zahl
    s: String, Standardwert '*'
    returns: String, der aus k-mal s besteht

    >>> zeichen(3)
    '***'

    >>> zeichen(2, 'ab')
    'abab'
    '''
    return k * s

help(zeichen)
```

PEP 257 empfiehlt drei doppelte Anführungszeichen. Wir benutzen im Kurs drei einfache, für Python ist das dasselbe.

Verändert eine Funktion ein Argument, muss das im Docstring stehen, denn das sieht man dem Aufruf nicht an.

```python
def entferne_erstes(a):
    '''
    a: nichtleere Liste
    verändert a: das erste Element wird entfernt
    returns: das entfernte Element
    '''
    return a.pop(0)

b = [7, 8, 9]
print(entferne_erstes(b), b)
```

Auch **Klassen**, ihre **Methoden** und ganze **Dateien** (Module) bekommen Docstrings. Der Docstring der Klasse steht unter
`class`, der eines Moduls in der ersten Zeile der Datei. Beginnt der Docstring mit einem kurzen Satz, was die Klasse oder
Methode tut, ist `help` besonders übersichtlich. `self` beschreiben wir nicht.

```python
class Konto:
    '''Ein Bankkonto, das nicht überzogen werden kann.'''

    def __init__(self, inhaber):
        '''
        inhaber: Name als String
        Der Kontostand ist zu Beginn 0.
        '''
        self.inhaber = inhaber
        self.stand = 0

    def einzahlen(self, betrag):
        '''
        Erhöht den Kontostand.
        betrag: positive Zahl
        '''
        self.stand += betrag

    def abheben(self, betrag):
        '''
        Vermindert den Kontostand, wenn das Geld reicht.
        betrag: positive Zahl
        returns: True, wenn abgehoben wurde, sonst False
        '''
        if betrag <= self.stand:
            self.stand -= betrag
            return True
        return False

help(Konto)
```

**Type Hints** (siehe Variablen) dokumentieren zusätzlich die Typen. Python selbst prüft sie nicht, sie sind ein Hinweis für Menschen
und Werkzeuge.

```python
def flaeche(breite: float, hoehe: float) -> float:
    return breite * hoehe

print(flaeche(2.5, 4))
```

#### Code-Stil nach PEP 8

Einheitlich formatierter Code ist leichter zu lesen. Die wichtigsten Regeln:

| Regel | so | nicht so |
|---|---|---|
| Einrückung mit 4 Leerzeichen | `    return x` | `  return x` |
| Leerzeichen um Zuweisung und Vergleich | `x = a + 1` | `x=a+1` |
| Leerzeichen nach dem Komma, nicht davor | `f(a, b)` | `f(a ,b)` |
| kein Leerzeichen innerhalb der Klammern | `f(a)` | `f( a )` |
| kein Leerzeichen vor dem Doppelpunkt | `def f(a):` | `def f(a) :` |
| Standardwert ohne Leerzeichen | `def f(a, b=0):` | `def f(a, b = 0):` |
| zwei Leerzeilen um Funktionen und Klassen in einer Datei | | |
| eine Leerzeile zwischen den Methoden einer Klasse | | |
| höchstens 79 Zeichen in einer Zeile | | |
| `import` am Anfang der Datei | | |

In unseren Notebooks stehen die Funktionen oft dichter beieinander, weil jede Zelle nur ein kurzes Beispiel enthält.

```python
# so nicht
def SUMME( a,b ) :
  S=a+b
  return S

# so
def summe(a, b):
    return a + b

print(SUMME(2, 3), summe(2, 3))
```

#### Zusammenfassung

- Gute Namen zuerst: Was der Name sagt, braucht keinen Kommentar.
- Ein Kommentar erklärt das **Warum**, nicht das Was, und wird mit dem Code geändert.
- Ein Docstring beschreibt, **wie** man eine Funktion oder Klasse benutzt: Argumente, Rückgabewert, veränderte Argumente, Beispiele.
- PEP 8 regelt Namen und Formatierung, PEP 257 die Docstrings.
- In Thonny kommentiert **Strg + 3** markierte Zeilen aus, in Jupyter **Strg + /**. Der Assistent von Thonny prüft den Code.
