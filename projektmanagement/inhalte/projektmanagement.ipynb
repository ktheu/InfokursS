## Projektmanagement

Größere Programme entstehen im Team. Damit das gelingt, muss ein Projekt geplant, durchgeführt und abgeschlossen werden. Wir lernen die wichtigsten Begriffe kennen und führen danach
selbst ein kleines Softwareprojekt durch.

### 1. Was ist ein Projekt?

Ein **Projekt** ist ein einmaliges Vorhaben mit einem klaren Ziel, einem festen Zeitrahmen und begrenzten Mitteln. Meistens arbeitet ein Team daran.

| Begriff | Bedeutung |
|---|---|
| Auftraggeber | möchte das Ergebnis haben, zum Beispiel eine Firma oder die Lehrkraft |
| Auftragnehmer | führt das Projekt durch, zum Beispiel unser Team |
| Projektauftrag | beschreibt Ziel, Zeitrahmen und Beteiligte des Projekts |
| Anforderungen | beschreiben, was das Ergebnis können muss |
| Prototyp | frühe, lauffähige Version, die wichtige Funktionen zeigt |
| Abnahme | Der Auftraggeber prüft das Ergebnis und nimmt es an. |

Die wöchentliche Hausaufgabe ist kein Projekt. Eine Abizeitung ist eines: Sie ist einmalig, hat ein klares Ziel und einen festen Termin.

### 2. Projektphasen

Jedes Projekt durchläuft typische **Phasen**:

| Phase | Was passiert? | Ergebnis |
|---|---|---|
| Planung | Ziel klären, Anforderungen sammeln, Aufgaben verteilen, Zeitplan erstellen | Projektauftrag, Anforderungen, Zeitplan |
| Entwurf | Aufbau des Programms festlegen: Klassen, Oberfläche, Datenbank | Projektentwurf |
| Umsetzung | programmieren und testen | Prototyp, Programm |
| Abschluss | Ergebnis vorführen und dokumentieren | Abnahme, Dokumentation |

### 3. Vorgehensmodelle

Ein **Vorgehensmodell** legt fest, in welcher Reihenfolge die Phasen ablaufen.

**Wasserfallmodell:** Die Phasen laufen einmal nacheinander ab. Eine Phase beginnt erst, wenn die vorherige abgeschlossen ist.

```
Planung -> Entwurf -> Umsetzung -> Test -> Abnahme
```

**Iteratives, agiles Vorgehen:** Das Projekt läuft in kurzen Runden ab (**Iterationen**). Nach jeder Runde gibt es ein lauffähiges Zwischenergebnis, das mit dem Auftraggeber besprochen wird.
Danach wird die nächste Runde geplant.

```
planen -> umsetzen -> testen -> vorführen -+
  ^                                        |
  +----------------------------------------+
```

| | Wasserfallmodell | agiles Vorgehen |
|---|---|---|
| Ablauf | einmal nacheinander | viele kurze Runden |
| Planung | am Anfang vollständig | in jeder Runde neu |
| Anforderungen | stehen am Anfang fest | dürfen sich ändern |
| Auftraggeber sieht Ergebnisse | am Ende | nach jeder Runde |
| geeignet, wenn | die Anforderungen klar und stabil sind | die Anforderungen noch unklar sind |

### 4. Anforderungen beschreiben

Anforderungen sollen so genau sein, dass man am Ende prüfen kann, ob sie erfüllt sind. „Das Programm soll schnell sein“ ist schlecht, „Die Suche liefert das Ergebnis in unter einer Sekunde“ ist gut.

- Das **Lastenheft** beschreibt, **was** der Auftraggeber haben möchte.
- Das **Pflichtenheft** beschreibt, **wie** der Auftragnehmer die Anforderungen umsetzt.
- **User Stories** beschreiben Anforderungen aus der Sicht der Benutzer:

```
Als <Rolle> möchte ich <Funktion>, damit <Nutzen>.
```

Beispiel: *Als Schülerin möchte ich falsch beantwortete Vokabeln wiederholen, damit ich gezielt üben kann.*

Aus den Anforderungen ergeben sich die **Systemfunktionen**, also das, was das Programm können muss, zum Beispiel „Vokabel hinzufügen“ oder „Abfrage starten“.

### 5. Ein Projektentwurf

Beispiel: ein Vokabeltrainer.

**Projektauftrag:** Ein Team aus drei Personen erstellt in sechs Wochen einen Vokabeltrainer mit grafischer Oberfläche.

**Anforderungen als User Stories:**

| Nr. | User Story | wichtig? |
|---|---|---|
| 1 | Als Schüler möchte ich Vokabeln eingeben, damit ich meine eigenen Listen habe. | muss |
| 2 | Als Schüler möchte ich abgefragt werden, damit ich lernen kann. | muss |
| 3 | Als Schüler möchte ich meine Trefferquote sehen, damit ich meinen Fortschritt erkenne. | muss |
| 4 | Als Schüler möchte ich falsche Vokabeln wiederholen, damit ich gezielt üben kann. | kann |

**Entwurf:** Die Vokabeln liegen in einer Datenbank. Die Klasse `Vokabel` speichert Wort, Übersetzung und Treffer. Die Oberfläche hat einen Knopf je Systemfunktion.

**Zeitplan und Aufgabenverteilung:**

| Woche | Ziel | Wer? |
|---|---|---|
| 1 | Anforderungen und Entwurf | alle |
| 2 bis 3 | Prototyp: Vokabeln eingeben und abfragen | Datenbank: Ali, Oberfläche: Bea, Abfrage: Cem |
| 4 bis 5 | Trefferquote, testen, Fehler beheben | alle |
| 6 | Dokumentation und Vorführung | alle |

Der **Prototyp** enthält nur die wichtigsten Funktionen. Er zeigt früh, ob der Entwurf trägt.

### 6. Unser Projekt

Wir führen in Teams ein eigenes Softwareprojekt durch und nutzen dabei Inhalte aus dem Kurs. Mögliche Ideen:

- ein Programm mit Datenbank, zum Beispiel eine Verwaltung für eine Bibliothek
- ein Spiel, dessen Ablauf wir mit einem Automaten modellieren
- ein Messenger-Modell, das Nachrichten verschlüsselt

**Das geben wir ab:**

1. Projektentwurf: Projektauftrag, Anforderungen (Lastenheft oder User Stories), Entwurf, Zeitplan, Aufgabenverteilung
2. Prototyp: lauffähiges Programm mit den wichtigsten Funktionen
3. Dokumentation: kurze Beschreibung und kommentierter Quellcode
4. Vorführung: Wir zeigen das Ergebnis und prüfen gemeinsam die Anforderungen (Abnahme).
