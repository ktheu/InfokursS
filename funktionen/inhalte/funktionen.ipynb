In einer Funktion fassen wir einen Codeabschnitt, der eine bestimmte Aufgabe erfüllt, zu einer Einheit zusammen
und geben ihm einen Namen. Funktionen kennen wir aus dem Brückenkurs. Hier ist das Wichtigste kompakt zusammengefasst.

### Begriffe

| Begriff | Bedeutung | im Beispiel |
|---|---|---|
| Funktionsdefinition | legt fest, was die Funktion tut, wird mit `def` eingeleitet | die Zeilen 1 bis 3 |
| Funktionskopf | Name und Parameterliste | `def calc(x, y):` |
| Funktionsrumpf | die eingerückten Anweisungen | `z = x + y` ... |
| Parameter | Variable in der Parameterliste, bekommt beim Aufruf einen Wert | `x`, `y` |
| Argument | konkreter Wert, der beim Aufruf übergeben wird | `2`, `4` |
| Rückgabewert | Ergebnis, das mit `return` zurückgegeben wird | `19` |
| Aufruf | Ausführen der Funktion | `calc(2, 4)` |

Beim Aufruf werden den Parametern die Argumente zugewiesen, hier also `x = 2` und `y = 4`. Erst dann wird der
Funktionsrumpf ausgeführt. Die Definition allein führt noch nichts aus.

````python
def calc(x, y):
    z = x + y
    return 3 * z + 1

a = calc(2, 4)         # Ergebnis in einer Variablen auffangen
print(a)
print(calc(2, 4))      # oder direkt weiterverwenden
print(calc(1, 1) + calc(0, 0))
````

In Jupyter-Notebooks wird der Wert der letzten Zeile einer Zelle auch ohne `print` angezeigt.

````python
calc(2, 4)
````

### Mit und ohne Parameter, mit und ohne Rückgabewert

Eine Funktion kann Parameter haben oder nicht, und sie kann einen Wert zurückgeben oder nicht.
Daraus ergeben sich vier Fälle:

| | ohne Rückgabewert | mit Rückgabewert |
|---|---|---|
| **ohne Parameter** | `def linie(): print('-----')` | `def wuerfeln(): return random.randint(1, 6)` |
| **mit Parameter** | `def linie(k): print(k * '-')` | `def quadrat(x): return x * x` |

Auch beim Aufruf einer Funktion ohne Parameter braucht man die Klammern: `linie()`. Ohne Klammern wird
die Funktion nicht aufgerufen.

````python
import random

def linie():
    print('-----')

def wuerfeln():
    return random.randint(1, 6)

def quadrat(x):
    return x * x

linie()
print(wuerfeln())
print(quadrat(7))
````

Wenn eine Funktion einen Parameter verlangt und wir beim Aufruf kein Argument übergeben, gibt es einen Fehler.

````python
quadrat()          # TypeError: missing 1 required positional argument: 'x'
````

##### `return` oder `print`?

Eine Funktion ohne `return` (oder mit einem `return` ohne Wert) gibt den besonderen Wert `None` zurück.
Ein häufiger Fehler: Die Funktion **gibt aus** statt **zurückzugeben**. Dann sieht man das Ergebnis zwar auf dem
Bildschirm, kann aber nicht damit weiterrechnen.

````python
def doppelt_print(x):
    print(2 * x)        # gibt aus

def doppelt_return(x):
    return 2 * x        # gibt zurück

a = doppelt_print(5)    # 10 erscheint auf dem Bildschirm ...
print(a)                # ... aber a ist None

b = doppelt_return(5)
print(b + 1)
````

````python
doppelt_print(5) + 1    # TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'
````

### Signatur

Name und Parameterliste einer Funktion bilden ihre **Signatur**. Sie legt fest, wie man die Funktion
aufrufen muss. In Java gehören zur Signatur auch die Datentypen der Parameter, und vor dem Namen steht der Datentyp
des Rückgabewerts. `void` bedeutet: kein Rückgabewert.

```java
static int quadrat(int x) {         // Rückgabetyp int, ein Parameter vom Typ int
    return x * x;
}

static void linie(int k) {          // kein Rückgabewert
    System.out.println("-".repeat(k));
}
```

In Java heißen Funktionen **Methoden**.

In Python kann man die Datentypen als **Typhinweise** angeben. Python prüft sie nicht, aber sie machen klar, was die
Funktion erwartet und was sie liefert.

````python
def zeichen(k: int, s: str) -> str:
    return k * s

print(zeichen(3, 'ab'))
print(zeichen(3, 7))          # kein Fehler: Typhinweise werden nicht geprüft
````

### Optionale Parameter

Ein optionaler Parameter hat einen **Standardwert** (default). Übergibt man beim Aufruf kein Argument dafür, wird
der Standardwert genommen. Optionale Parameter müssen am Ende der Parameterliste stehen.

````python
def zeichen(k, s='*'):
    return k * s

print(zeichen(5, 'abi'))
print(zeichen(20))            # s bekommt den Standardwert '*'
````

##### Schlüsselwortargumente

Beim Aufruf kann man ein Argument auch mit dem Namen des Parameters übergeben: `zeichen(k=5, s='-')`.
Dann spielt die Reihenfolge keine Rolle, und der Aufruf ist oft besser lesbar. Bei `print` kennen wir das schon:
`sep` und `end` sind optionale Parameter.

````python
print(zeichen(s='-', k=10))

print(1, 2, 3, sep=' | ')
print('ohne Zeilenumbruch', end=' ... ')
print('weiter')
````

### Mehrere `return`-Anweisungen

Eine Funktion kann mehrere `return`-Anweisungen haben. Beim ersten `return`, das ausgeführt wird, wird die Funktion
**sofort verlassen**. Ein `else` ist dann oft überflüssig.

````python
def temperatur(k):
    if k >= 30:
        return 'heiß'
    if k >= 15:
        return 'warm'
    return 'kalt'

print(temperatur(45), temperatur(20), temperatur(3))
````

Das ist besonders praktisch bei der **Suche**: Sobald man gefunden hat, was man sucht, gibt man das Ergebnis zurück.
Die Schleife muss nicht weiterlaufen. Erst wenn die Schleife ganz durchgelaufen ist, steht fest, dass es nichts gibt.

````python
def ist_primzahl(n):
    if n < 2:
        return False
    for t in range(2, n):
        if n % t == 0:
            return False        # Teiler gefunden, fertig
    return True                 # kein Teiler gefunden

def erste_negative(zahlen):
    for i in range(len(zahlen)):
        if zahlen[i] < 0:
            return i
    return -1                   # nicht gefunden

print(ist_primzahl(17), ist_primzahl(21))
print(erste_negative([3, 8, -2, 5, -7]), erste_negative([1, 2]))
````

### Docstring

Der *Docstring* steht direkt unter dem Funktionskopf. Er beschreibt, welche Argumente die Funktion erwartet und was
sie zurückgibt. Mit `help` kann man ihn anzeigen.

````python
def zeichen(k, s='*'):
    '''
    k: positive ganze Zahl
    s: String, Standardwert '*'
    returns: String, der aus k-mal s besteht
    '''
    return k * s

help(zeichen)
````

### Funktionen rufen Funktionen auf

Eine Funktion kann andere Funktionen aufrufen. So zerlegt man ein größeres Problem in kleine Teilprobleme,
die man einzeln lösen und testen kann. Jede Funktion erledigt genau eine Aufgabe.

````python
def ist_primzahl(n):
    if n < 2:
        return False
    for t in range(2, n):
        if n % t == 0:
            return False
    return True

def primzahlen_bis(n):
    ergebnis = []
    for k in range(2, n + 1):
        if ist_primzahl(k):
            ergebnis.append(k)
    return ergebnis

def anzahl_primzahlen(von, bis):
    return len(primzahlen_bis(bis)) - len(primzahlen_bis(von - 1))

print(primzahlen_bis(30))
print(anzahl_primzahlen(10, 30))
````

### Wie wir Funktionen schreiben

Für die meisten Fälle gilt: **Funktionen sollten alles, was sie für ihre Arbeit benötigen, über Parameter erhalten
und alle Ergebnisse mit `return` zurückgeben.** Wir verwenden in der Regel kein `input` und kein `print` innerhalb von
Funktionen. Ausnahme: Funktionen, deren einziger Zweck die Ein- oder Ausgabe ist.

Nur dann kann man die Funktion in anderen Programmen wiederverwenden und automatisch testen.

````python
# SO NICHT:
def ggt():
    a = int(input('Bitte eine positive ganze Zahl eingeben: '))
    b = int(input('Bitte noch eine positive ganze Zahl eingeben: '))
    while a != b:
        if a > b:
            a = a - b
        else:
            b = b - a
    print(f'Der größte gemeinsame Teiler beider Zahlen ist {a}.')
````

````python
# SONDERN SO:
def ggt(a, b):
    '''
    a, b: positive ganze Zahlen
    returns: größter gemeinsamer Teiler von a und b
    '''
    while a != b:
        if a > b:
            a = a - b
        else:
            b = b - a
    return a

a = int(input('Bitte eine positive ganze Zahl eingeben: '))
b = int(input('Bitte noch eine positive ganze Zahl eingeben: '))
print(f'Der größte gemeinsame Teiler beider Zahlen ist {ggt(a, b)}.')
````

### Lokale und globale Variablen

Den Gültigkeitsbereich von Variablen haben wir im Notebook zu Variablen besprochen. Kurz zusammengefasst:

- Parameter und Variablen, die in einer Funktion einen Wert bekommen, sind **lokal**. Sie existieren nur, solange
  die Funktion läuft. Beim Aufruf entsteht dafür ein eigener Bereich (*frame*), das lässt sich gut im
  [Python Tutor](https://pythontutor.com/visualize.html#mode=edit) verfolgen.
- **Globale** Variablen kann eine Funktion lesen, solange sie nicht von einer gleichnamigen lokalen Variablen
  verdeckt werden.
- Um eine globale Variable in einer Funktion zu **verändern**, muss man sie mit `global` deklarieren.
  Das sollte man aber vermeiden: Funktionen sollten nur mit ihren Parametern arbeiten.

````python
x = 10                 # globale Variable

def lesen():
    return x + 2       # globale Variable wird gelesen

def verdecken(x):
    return x + 2       # der Parameter x verdeckt die globale Variable

def aendern():
    global x
    x = 7              # verändert die globale Variable

print(lesen(), verdecken(5), x)
aendern()
print(x)
````

### Zusammenfassung

- Funktionen sollten alles, was sie für ihre Arbeit benötigen, über Parameter erhalten und alle Ergebnisse mit
  `return` zurückgeben.
- Beim Aufruf werden den Parametern die Argumente zugewiesen. Optionale Parameter haben einen Standardwert und stehen
  am Ende der Parameterliste. Mit Schlüsselwortargumenten `name=wert` kann man Argumente auch über den Namen übergeben.
- Name und Parameterliste bilden die Signatur der Funktion. In Java gehören die Datentypen und der Rückgabetyp dazu.
- Eine Funktion ohne `return` gibt `None` zurück. `print` in einer Funktion ist kein Rückgabewert.
- Beim ersten ausgeführten `return` wird die Funktion sofort verlassen.
- Parameter und Variablen, die in der Funktion einen Wert bekommen, sind lokal. Globale Variablen kann man lesen,
  zum Verändern braucht man `global`, das sollte man aber vermeiden.
- Der Docstring beschreibt die Funktion.

Wir schreiben Funktionen, weil

- wir einzelne Codeabschnitte besser testen können,
- Programme mit sprechenden Funktionsnamen besser lesbar werden,
- wir die Codeabschnitte an anderen Stellen wiederverwenden können,
- wir ein großes Problem in kleine Teilprobleme zerlegen können.
