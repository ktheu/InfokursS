#### LAN  

Ein **LAN** (Local Area Network) ist auf kleine räumliche Umgebungen beschränkt (z.B. Schule oder Firma). Auch ein Heimnetzwerk ist ein LAN.

In einem lokalen Netzwerk gibt drei Komponenten: 
- **Endgeräte** - Ein Endgerät ist der eigentliche Kommunikationspartner im Netz – es sendet und empfängt Daten. Beispiele: PC, Laptop, Smartphone, Tablet, Drucker, Server.
- **Verbindungen** - Eine Verbindung überträgt die Daten physisch zwischen den Geräten: Netzwerkkabel, Glasfaser, WLAN  
- **Verteiler**
Ein Verteiler verbindet mehrere Geräte miteinander und leitet die Datenpakete weiter: Switch, Router, Access Point, Modem






<img src='./lan1.png' width='801'>

Bei einem Heimnetzwerk verbindet der **Router** alle Geräte, die zum LAN gehören, mit dem Internet. Der Router ist über ein **Modem** mit dem Internetdienstanbieter (ISP Internet Service Provider) verbunden. Reichen die Anschlüsse des Routers nicht aus, benötigt man einen Verteiler, z.B. einen **Switch**. Switche, mit denen sich Geräte per Funk verbinden lassen, nennt man Wireless Access Point, kurz **AP**.
Moderne Router haben meist das Modem, einige LAN-Anschlüsse und einen AP eingebaut.

In Routern ist meist eine Software eingebaut, die zwischen Internet und dem lokalen Netzwerk eine Kontrollstelle aufbaut (**Firewall**).

Das **Internet** ist ein Netz von Netzen.

#### Eindeutige Adressierung
Damit ein Rechner im Internet gefunden wird, benötigt er eine eindeutige Adresse. Das Prinzip der eindeutigen Adressierung findet sich auch in anderen Bereichen:


- Handynummer: eine SIM-Karte - Ländervorwahl + Rufnummer
- E-Mail-Adresse: ein Postfach - Nutzername + Domain
- Postanschrift	ein Haushalt - Land + PLZ + Straße + Hausnummer

In allen Fällen gilt dasselbe Prinzip: Ein eindeutige Adresse sorgt dafür, dass eine Nachricht genau bei der richtigen Person oder dem richtigen Gerät ankommt – und nicht woanders.


#### IP-Adressen

Die IPv4-Adresse ist eine 4 Byte lange Zahl. Die Zahlen werden jeweils durch einen Punkt getrennt, z.B:

    192.168.123.137

Es gibt $(2^8)^4 = 2^{32}$ verschiedene IPv4 Adressen (ca. 4 Milliarden).

Eine IPv6 Adresse verwendet 16 Byte und sieht in Hex-Darstellung z.B. so aus:

    4003:0dc8:15a6:08d4:2319:3b2a:0040:3221

Wir betrachten im Folgenden nur IPv4.
Zur einer IP-Adresse gehört eine Subnetz-Maske, die die Adresse in einen Netz und einen Host-Anteil aufteilt. 

    IP-Adresse:       192.168.123.137  =	11000000.10101000.01111011.10001001
    Subnetz-Maske:    255.255.255.000  =	11111111.11111111.11111111.00000000
    Netzwerk-Kennung: 192.168.123.000  =    11000000.10101000.01111011.00000000
    Computer-Kennung:             137

Das Ende der Einsen in der Subnetz-Maske kann auch mitten im Byte liegen. Geräte mit derselben Netzwerkkennung liegen im selben Netz. 
Die Subnetz-Maske kann ersetzt werden durch die Angabe eines Suffix, der die Anzahl der Einsen in der Subnetz-Maske angibt. Die obige
IP-Adresse könnte man damit auch schreiben als 192.168.123.137/24.

##### Beispiel: Liegen zwei Geräte im selben Netz?

```
Gegeben: ip1 = 10.0.5.100, ip2 = 10.0.5.130, ip3 = 10.0.5.120
Subnetzmaske für alle drei: 255.255.255.224
```

**Weg 1: Netzgröße berechnen**

Die Subnetzmaske endet auf 224 = 11100000. Im letzten Byte gehören also die ersten 3 Bit zum Netzteil und die letzten 5 Bit zum Geräteteil. Mit 5 Bit gibt es $2^5 = 32$ verschiedene Werte (00000 bis 11111), jedes Netz umfasst also 32 Adressen.

Schneller geht es mit 256 - 224 = 32: Die 224 ist die Summe der Stellenwerte der Netz-Bits (128 + 64 + 32). Was bis 256 fehlt, sind genau die 32 Möglichkeiten für die Geräte-Bits.

Die Netze beginnen deshalb immer bei einem Vielfachen von 32: bei 0, 32, 64, 96, 128, 160, ...

- ip1 = 10.0.5.100 liegt im Netz 10.0.5.96 bis 10.0.5.127.
- ip2 = 10.0.5.130 liegt im Netz 10.0.5.128 bis 10.0.5.159.
- ip3 = 10.0.5.120 liegt im Netz 10.0.5.96 bis 10.0.5.127.

**Weg 2: Binärdarstellung**

224 = 11100000, also eine /27-Maske. Im letzten Byte gehören die ersten 3 Bit zum Netzteil.

```
ip1:  100 = 011|00100
ip2:  130 = 100|00010
ip3:  120 = 011|11000
```

Ergebnis: ip1 und ip3 haben dieselben Netz-Bits (011), sie liegen im selben Netz und können direkt
miteinander kommunizieren. ip2 hat andere Netz-Bits (100) und liegt in einem anderen Netz.
Für eine Verbindung von ip1 zu ip2 wird ein Router benötigt.

#### Standardgateway und DNS-Server

- Das **Standardgateway** ist die Adresse, an die ein Gerät Pakete für fremde Netze schickt.
- Der **DNS-Server** übersetzt leicht merkbare Namen von Webseiten in die passenden IP-Adressen. Wenn man zum Beispiel www.inf-schule.de eingibt, fragt der Computer einen DNS-Server nach der zugehörigen IP-Adresse. Erst mit dieser Adresse kann die Verbindung zur Webseite hergestellt werden. Deshalb wird das DNS oft als das „Telefonbuch des Internets“ bezeichnet.

#### Domainnamen

Ein **Domainname** wie `www.inf-schule.de` ist der merkbare Name eines Rechners im Internet. Er wird von **rechts nach links** gelesen:

- `de` ist die **Top-Level-Domain** (z.B. `de`, `com`, `org`),
- `inf-schule` ist die **Second-Level-Domain**, also der Name, den man registrieren lässt,
- `www` ist eine **Subdomain**, hier der Webserver der Domain.

Jeder Name muss weltweit eindeutig sein. Zum Aufrufen einer Seite wird der Name vom DNS-Server in die IP-Adresse übersetzt (siehe oben).

#### Routing-Problem

Im Internet sind viele Netzwerke miteinander verbunden. Oft können Daten nicht direkt vom Sender zum Empfänger gelangen, sondern müssen über mehrere Zwischenstationen weitergeleitet werden. Dabei stellt sich die Frage, welchen Weg die Daten nehmen sollen. Diese Aufgabe nennt man das Routing-Problem.

Wenn ein Datenpaket ankommt, entscheidet der Router anhand seiner Routing-Tabelle, an welchen nächsten Router das Paket weitergeleitet wird. So finden die Daten Schritt für Schritt ihren Weg zum Ziel, auch wenn dabei mehrere Router durchlaufen werden müssen.

#### Client-Server

Ein **Server** stellt in einem Netz einen Dienst zur Verfügung. Ein **Client** nimmt diesen Dienst in Anspruch.

Ein **Webserver** stellt HTML-Seiten zur Verfügung, der Client dazu ist ein **Browser**. <br>
Ein **DNS-Server** sorgt dafür, dass wir statt IP-Adressen Domainnamen verwenden können.
