## Filius - Übungen

**Filius** (Freie Interaktive Lernumgebung für Internetworking der Universität Siegen) ist eine kostenlose Lernsoftware, mit der man Rechnernetze am Computer aufbauen und testen kann.

- Download: [www.lernsoftware-filius.de](https://www.lernsoftware-filius.de)
- Ausführliche Anleitungen: [Einführung in Filius](https://www.lernsoftware-filius.de/downloads/Einfuehrung_Filius.pdf), [Skriptum Netzwerktechnik mit Filius](https://www.lernsoftware-filius.de/downloads/Skriptum_Netzwerk_Filius_2026.pdf), [Filius-Aufgaben](https://www.lernsoftware-filius.de/downloads/karmann/Filiusaufgaben.pdf)

In den folgenden Übungen baust du Schritt für Schritt immer größere Netze auf:

| Übung | Thema |
|:-:|:--|
| 1 | Zwei Notebooks direkt verbinden |
| 2 | Mehrere Notebooks über einen Switch verbinden |
| 3 | Webserver und Webbrowser |
| 4 | DNS-Server: Webseite über einen Namen aufrufen |
| 5 | Eigene Webseiten mit Links |
| 6 | Zwei Netze über einen Vermittlungsrechner (Router) verbinden |

### Grundlagen

#### Die Modi

| Modus | Symbol | Was macht man dort? |
|:--|:--|:--|
| **Entwurfsmodus** | Hammer | Geräte einfügen, mit Kabeln verbinden und konfigurieren (Name, IP-Adresse, Netzmaske, Gateway, ...) |
| **Aktionsmodus** | grüner Pfeil | Das Netz „läuft“. Software installieren und starten, Verbindungen testen (z. B. mit `ping`) |
| Dokumentationsmodus | Bleistift | Textfelder und Flächen zur Beschriftung einfügen |

Filius startet immer im Entwurfsmodus. Ein Netz kann im Entwurfsmodus richtig aussehen und im Aktionsmodus trotzdem nicht funktionieren, z. B. wegen eines Tippfehlers in einer IP-Adresse.

#### Die Bauteile

| Bauteil | Wofür? |
|:--|:--|
| **Notebook** | für Clients, also „normale“ Rechner der Benutzer |
| **Rechner** | für Server (Webserver, DNS-Server) |
| **Kabel** | verbindet zwei Bauteile |
| **Switch** | verbindet mehrere Geräte **im selben Netz**. Ein Switch braucht keine IP-Adresse. |
| **Vermittlungsrechner** | der Router: verbindet **verschiedene Netze** miteinander |

Notebook und Rechner funktionieren in Filius gleich. Die Unterscheidung hilft nur, das Netz übersichtlich zu halten.

#### Die wichtigsten Handgriffe

| Was? | Wie? | Modus |
|:--|:--|:--|
| Gerät einfügen | Bauteil aus der linken Leiste auf die Arbeitsfläche ziehen | Entwurf |
| Geräte verbinden | Kabel anklicken, dann nacheinander die beiden Geräte anklicken | Entwurf |
| Gerät konfigurieren | Doppelklick auf das Gerät: Name, IP-Adresse, Netzmaske, Gateway, Domain Name Server | Entwurf |
| Software installieren | Doppelklick auf das Gerät öffnet den Desktop → *Software-Installation* → Programm auswählen → Pfeil nach links → *Änderungen übernehmen* | Aktion |
| Software starten | Auf dem Desktop des Geräts auf das Programmsymbol klicken | Aktion |
| Datenverkehr ansehen | Rechtsklick auf das Gerät → *Datenaustausch anzeigen* | Aktion |
| Speichern | *Speichern* in der oberen Leiste (Dateiendung `.fls`) | beide |

**Tipps**

- Benenne die Geräte nach ihrer Aufgabe und hänge das Ende der IP-Adresse an, z. B. `Webserver_20` für die Adresse 192.168.5.20.
- Wir verwenden für alle Geräte die Netzmaske **255.255.255.0**. Unsere Heimnetzadressen beginnen mit **192.168**.
- Im Aktionsmodus öffnet ein Doppelklick auf ein Gerät dessen Desktop in einem eigenen Fenster. Schiebe diese Fenster neben das Filius-Hauptfenster (notfalls das Hauptfenster etwas verkleinern). So siehst du gleichzeitig die Befehlszeile und die grün leuchtenden Kabel im Netz.
- Bei jedem Datenaustausch leuchten die Kabel grün. So siehst du, welchen Weg die Daten nehmen.

#### So beschreibst du eine Filius-Konfiguration

In den Aufgaben sollst du eine Konfiguration skizzieren und beschreiben, wie getestet wird. Dazu gehören:

1. **Geräte und Verbindungen**: welche Notebooks, Rechner, Switches und Vermittlungsrechner, und wie sie verbunden sind.
2. **Einstellungen** an jedem Gerät: IP-Adresse und Netzmaske, wenn nötig Gateway und Domain Name Server.
3. **Software**: welche Programme auf welchem Gerät installiert und gestartet werden.
4. **Test**: womit du prüfst, dass alles funktioniert, z. B. `ping 192.168.5.11` oder Aufruf einer Webseite im Webbrowser.

Eine Skizze kann ganz einfach sein:

```
Netzmaske für alle Geräte: 255.255.255.0

NB_10 ---------------- NB_11
192.168.5.10           192.168.5.11

Test auf NB_10: ping 192.168.5.11
```

### Übung 1: Zwei Notebooks direkt verbinden

**Ziel:** Zwei Notebooks sollen über ein Kabel Daten austauschen.

| Gerät | Name | IP-Adresse | Netzmaske |
|:--|:--|:--|:--|
| Notebook | NB_10 | 192.168.5.10 | 255.255.255.0 |
| Notebook | NB_11 | 192.168.5.11 | 255.255.255.0 |

**Im Entwurfsmodus**

1. Zwei Notebooks auf die Arbeitsfläche ziehen.
2. Mit dem Kabel verbinden.
3. Mit Doppelklick auf jedes Notebook Name, IP-Adresse und Netzmaske eintragen.

**Im Aktionsmodus**

4. Auf NB_10 die Software **Befehlszeile** installieren und starten.
5. Testen mit dem Befehl `ping 192.168.5.11`. Es kommen vier Antworten zurück.
6. Mit `ipconfig` zeigt die Befehlszeile die eigenen Einstellungen an.

**Beobachten und ausprobieren**

- Rechtsklick auf NB_10 → *Datenaustausch anzeigen*. Vor den Ping-Paketen fragt NB_10 zuerst nach der MAC-Adresse von NB_11 (Protokoll ARP).
- Ändere im Entwurfsmodus die Adresse von NB_11 auf 192.168.**6**.11. Was passiert jetzt beim Ping? Warum?

<details>
<summary>Skizze und Test anzeigen</summary>

```
Netzmaske für alle Geräte: 255.255.255.0

NB_10 ---------------- NB_11
192.168.5.10           192.168.5.11
Befehlszeile

Test auf NB_10: ping 192.168.5.11

Mit 192.168.6.11 klappt der Ping nicht: NB_11 liegt dann in einem
anderen Netz (Netzteil 192.168.6 statt 192.168.5). Dafür bräuchte man
einen Router, siehe Übung 6.
```

</details>

### Übung 2: Mehrere Notebooks über einen Switch

**Ziel:** Drei Notebooks im selben Netz sollen sich gegenseitig erreichen. Mit einem Kabel lassen sich nur zwei Geräte verbinden, deshalb brauchen wir einen **Switch**.

| Gerät | Name | IP-Adresse | Netzmaske |
|:--|:--|:--|:--|
| Notebook | NB_10 | 192.168.5.10 | 255.255.255.0 |
| Notebook | NB_11 | 192.168.5.11 | 255.255.255.0 |
| Notebook | NB_12 | 192.168.5.12 | 255.255.255.0 |
| Switch | Switch | – | – |

**Im Entwurfsmodus**

1. Drei Notebooks und einen Switch einfügen.
2. Jedes Notebook mit einem Kabel mit dem Switch verbinden.
3. Die Notebooks konfigurieren. Der Switch bekommt **keine** IP-Adresse.

**Im Aktionsmodus**

4. Auf NB_10 die Befehlszeile installieren.
5. Testen mit `ping 192.168.5.11` und `ping 192.168.5.12`.

**Beobachten und ausprobieren**

- Achte darauf, welche Kabel beim Ping grün leuchten. Beim ersten Ping fragt NB_10 per Broadcast an alle Geräte nach der MAC-Adresse. Danach schickt der Switch die Pakete nur noch an das richtige Gerät.

<details>
<summary>Skizze und Test anzeigen</summary>

```
Netzmaske für alle Geräte: 255.255.255.0

NB_10 192.168.5.10 ----\
                         \
NB_11 192.168.5.11 ------ Switch (keine IP-Adresse)
                         /
NB_12 192.168.5.12 ----/

Test auf NB_10: ping 192.168.5.11
                ping 192.168.5.12
```

</details>

### Übung 3: Webserver und Webbrowser

**Ziel:** Ein Notebook soll die Webseite eines Webservers anzeigen.

| Gerät | Name | IP-Adresse | Netzmaske | Software |
|:--|:--|:--|:--|:--|
| Notebook | NB_10 | 192.168.5.10 | 255.255.255.0 | Webbrowser |
| Rechner | Webserver_20 | 192.168.5.20 | 255.255.255.0 | Webserver |
| Switch | Switch | – | – | – |

Für zwei Geräte würde auch ein Kabel reichen. Mit dem Switch können wir das Netz in Übung 4 leicht erweitern.

**Im Entwurfsmodus**

1. Notebook, Rechner und Switch einfügen und verbinden, Adressen eintragen.

**Im Aktionsmodus**

2. Auf Webserver_20 die Software **Webserver** installieren, öffnen und auf **Starten** klicken.
3. Auf NB_10 den **Webbrowser** installieren und starten.
4. Im Webbrowser die Adresse `http://192.168.5.20` eingeben. Die Startseite des Webservers erscheint.

**Ausprobieren**

- Auf Webserver_20 den **Text-Editor** installieren. Über *Datei → Öffnen* im Ordner `webserver` die Datei `index.html` öffnen, die Überschrift ändern und speichern. Im Webbrowser die Seite neu laden.
- Webserver stoppen und die Seite neu laden. Was passiert?

<details>
<summary>Skizze und Test anzeigen</summary>

```
Netzmaske für alle Geräte: 255.255.255.0

NB_10 --------- Switch --------- Webserver_20
192.168.5.10                     192.168.5.20
Webbrowser                       Webserver installiert und gestartet

Test: Webbrowser auf NB_10 mit der Adresse http://192.168.5.20
      Die Startseite des Webservers wird angezeigt.
```

</details>

### Übung 4: DNS-Server

**Ziel:** Die Webseite aus Übung 3 soll über den Namen `www.klasse10.de` erreichbar sein. Der **DNS-Server** übersetzt den Namen in die IP-Adresse des Webservers.

Wir erweitern das Netz aus Übung 3:

| Gerät | Name | IP-Adresse | Domain Name Server | Software |
|:--|:--|:--|:--|:--|
| Notebook | NB_10 | 192.168.5.10 | **192.168.5.30** | Webbrowser |
| Rechner | Webserver_20 | 192.168.5.20 | | Webserver |
| Rechner | DNS_30 | 192.168.5.30 | | DNS-Server |

**Im Entwurfsmodus**

1. Einen weiteren Rechner DNS_30 einfügen und mit dem Switch verbinden.
2. Bei NB_10 im Feld **Domain Name Server** die Adresse des DNS-Servers eintragen: 192.168.5.30. Sonst weiß das Notebook nicht, wen es nach dem Namen fragen soll.

**Im Aktionsmodus**

3. Auf DNS_30 die Software **DNS-Server** installieren und öffnen.
4. Als Domainname `www.klasse10.de` und als IP-Adresse `192.168.5.20` eintragen, auf **Hinzufügen** klicken. Der Eintrag erscheint in der Liste.
5. Auf **Starten** klicken.
6. Im Webbrowser auf NB_10 die Adresse `http://www.klasse10.de` eingeben.

**Beobachten und ausprobieren**

- Im Datenaustausch von NB_10 siehst du: Zuerst geht eine Anfrage an den DNS-Server, erst danach an den Webserver.
- In der Befehlszeile zeigt der Befehl `host www.klasse10.de` die IP-Adresse zum Namen.
- DNS-Server stoppen und die Seite über den Namen aufrufen. Funktioniert der Aufruf über die IP-Adresse noch?

<details>
<summary>Skizze und Test anzeigen</summary>

```
Netzmaske für alle Geräte: 255.255.255.0

                                 Webserver_20
                              /  192.168.5.20
NB_10 --------- Switch -------   Webserver gestartet
192.168.5.10                  \
DNS: 192.168.5.30                DNS_30
Webbrowser                       192.168.5.30
                                 DNS-Server gestartet mit dem Eintrag
                                 www.klasse10.de -> 192.168.5.20

Test: Webbrowser auf NB_10 mit der Adresse http://www.klasse10.de
```

</details>

### Übung 5: Eigene Webseiten mit Links

**Ziel:** Du erweiterst das Netz aus Übung 4 um einen zweiten Webserver. Die Seite des ersten Servers bekommt einen Link auf die Seite des zweiten Servers.

1. Einen zweiten **Webserver** mit einer freien IP-Adresse im selben Netz einfügen und mit dem Switch verbinden. Auf ihm die Software **Webserver** installieren und starten.
2. Auf dem DNS-Server einen zweiten Namen für diesen Server eintragen, zum Beispiel `www.schule.de`.
3. Auf beiden Servern mit dem **Text-Editor** die Datei `index.html` im Ordner `webserver` ändern, damit die Seiten verschieden aussehen.
4. In die Seite des ersten Servers einen Link einfügen:

```html
<a href="http://www.schule.de">Zur Schule</a>
```

5. Im Aktionsmodus im Webbrowser des Notebooks die erste Seite aufrufen und dem Link folgen.

### Übung 6: Zwei Netze über einen Vermittlungsrechner verbinden

**Ziel:** Zwei Netze mit je zwei Notebooks sollen miteinander verbunden werden. Ein Switch reicht dafür nicht, denn er verbindet nur Geräte im selben Netz. Wir brauchen einen **Vermittlungsrechner** (Router).

| Netz | Gerät | IP-Adresse | Gateway |
|:--|:--|:--|:--|
| 192.168.5.0 | NB_A10 | 192.168.5.10 | 192.168.5.1 |
| 192.168.5.0 | NB_A11 | 192.168.5.11 | 192.168.5.1 |
| 192.168.6.0 | NB_B10 | 192.168.6.10 | 192.168.6.1 |
| 192.168.6.0 | NB_B11 | 192.168.6.11 | 192.168.6.1 |
| beide | Vermittlungsrechner | Schnittstelle 1: 192.168.5.1<br>Schnittstelle 2: 192.168.6.1 | – |

Netzmaske überall 255.255.255.0.

**Im Entwurfsmodus**

1. Zwei Netze aufbauen: je zwei Notebooks an einem eigenen Switch (Switch_A und Switch_B).
2. Einen **Vermittlungsrechner** einfügen. Filius fragt nach der Anzahl der Schnittstellen: **2**.
3. Den Vermittlungsrechner mit Switch_A und mit Switch_B verbinden.
4. Doppelklick auf den Vermittlungsrechner: Für jede Schnittstelle gibt es eine eigene Registerkarte. Die Schnittstelle zu Switch_A bekommt 192.168.5.1, die zu Switch_B bekommt 192.168.6.1. Prüfe dabei, welche Schnittstelle mit welchem Switch verbunden ist.
5. Die Notebooks konfigurieren, zunächst **ohne** Gateway.

**Im Aktionsmodus**

6. Auf NB_A10 die Befehlszeile installieren und `ping 192.168.6.10` eingeben. Der Ping klappt **nicht**: NB_A10 erkennt, dass das Ziel in einem anderen Netz liegt, weiß aber nicht, wohin es das Paket schicken soll.

**Gateway eintragen**

7. Zurück im Entwurfsmodus bei jedem Notebook das **Gateway** eintragen. Das ist die Adresse der Router-Schnittstelle im eigenen Netz: 192.168.5.1 für Netz A, 192.168.6.1 für Netz B.
8. Im Aktionsmodus noch einmal `ping 192.168.6.10`. Jetzt kommen die Antworten an.

**Beobachten und ausprobieren**

- Achte auf die grün leuchtenden Kabel: Das Paket geht über Switch_A, den Vermittlungsrechner und Switch_B.
- Der Befehl `traceroute 192.168.6.10` zeigt die Stationen auf dem Weg zum Ziel.
- Trage bei NB_B10 ein falsches Gateway ein, z. B. 192.168.6.99. Was passiert beim Ping von NB_A10? Die Antwort von NB_B10 muss ja auch zurückfinden.

**Weiteres Szenario:** Was passiert, wenn die Geräte das Gateway nicht kennen? Lösche bei einem Notebook aus Netz A den Eintrag **Gateway** und teste `ping` zu einem Notebook in Netz B.
Innerhalb von Netz A funktioniert `ping` weiterhin, in das andere Netz aber nicht mehr, denn das Notebook weiß nicht, wohin es die Pakete für fremde Netze schicken soll.

<details>
<summary>Skizze und Test anzeigen</summary>

```
Netzmaske für alle Geräte: 255.255.255.0

Netz 192.168.5.0                    Netz 192.168.6.0

NB_A10  192.168.5.10                NB_B10  192.168.6.10
        Gateway 192.168.5.1                 Gateway 192.168.6.1
NB_A11  192.168.5.11                NB_B11  192.168.6.11
        Gateway 192.168.5.1                 Gateway 192.168.6.1
   |                                   |
Switch_A ---- Vermittlungsrechner ---- Switch_B
       192.168.5.1         192.168.6.1
       (Schnittstelle 1)   (Schnittstelle 2)

Test auf NB_A10: ping 192.168.6.10   (über die Netzgrenze hinweg)
```

</details>

### Fehlersuche: Es klappt nicht?

| Problem | Mögliche Ursache |
|:--|:--|
| Software lässt sich nicht installieren | Du bist noch im Entwurfsmodus. Software gibt es nur im Aktionsmodus. |
| IP-Adresse lässt sich nicht ändern | Du bist im Aktionsmodus. Einstellungen ändert man im Entwurfsmodus. |
| `ping` ins eigene Netz klappt nicht | Tippfehler in der IP-Adresse oder Netzmaske, Kabel fehlt |
| `ping` in ein anderes Netz klappt nicht | Gateway fehlt oder ist falsch, beim Sender **oder** beim Empfänger. Schnittstellen des Vermittlungsrechners falsch eingetragen |
| Webseite über die IP-Adresse erscheint nicht | Webserver nicht gestartet, falsche IP-Adresse im Webbrowser |
| Webseite über den Namen erscheint nicht | DNS-Server nicht gestartet, Eintrag fehlt oder hat die falsche IP-Adresse, beim Notebook kein Domain Name Server eingetragen |
