## Programme mit grafischer Oberfläche

Bisher haben unsere Programme mit `input` und `print` gearbeitet. Programme auf dem Computer oder dem Smartphone haben dagegen eine **grafische
Benutzeroberfläche** (GUI, *Graphical User Interface*) mit Fenstern, Knöpfen und Eingabefeldern. Solche Programme sind **interaktiv**: Sie warten darauf,
dass etwas passiert, und reagieren dann.

Wir benutzen dafür die Bibliothek `tkinter`, die zu Python gehört. Wir führen die Programme in **Thonny** aus.

1. Das erste Fenster
2. Ein Knopf löst eine Funktion aus
3. Text eingeben
4. Mit Zahlen rechnen
5. Beispiel: Zahlenraten

### 1. Das erste Fenster

Ein Fenster und alles, was darin steht, sind Objekte. Bedienelemente wie Beschriftungen und Knöpfe heißen **Widgets**.

```python
import tkinter as tk

fenster = tk.Tk()
fenster.title('Mein Programm')
fenster.geometry('300x120')

beschriftung = tk.Label(fenster, text='Hallo Welt')
beschriftung.pack()

fenster.mainloop()
```

| Zeile | Bedeutung |
|---|---|
| `import tkinter as tk` | lädt die Bibliothek, wir schreiben danach kurz `tk.` |
| `tk.Tk()` | erzeugt das Fenster |
| `fenster.title(...)`, `fenster.geometry(...)` | legen den Titel und die Größe in Pixeln fest |
| `tk.Label(fenster, text=...)` | erzeugt eine Beschriftung im Fenster |
| `beschriftung.pack()` | setzt die Beschriftung ins Fenster, erst dann ist sie zu sehen |
| `fenster.mainloop()` | zeigt das Fenster und wartet auf Aktionen, bis es geschlossen wird |

`mainloop()` steht immer am Ende. Die Zeilen dahinter laufen erst, wenn das Fenster geschlossen ist.

### 2. Ein Knopf löst eine Funktion aus

Ein `Button` bekommt mit `command=` die Funktion, die beim Klicken aufgerufen wird. Hinter dem Namen stehen dabei **keine Klammern**:
Wir übergeben die Funktion, wir rufen sie nicht auf.

Mit `beschriftung['text'] = ...` ändern wir den angezeigten Text.

```python
import tkinter as tk

def klick():
    beschriftung['text'] = 'Danke!'

fenster = tk.Tk()
fenster.title('Knopf')
fenster.geometry('300x120')

beschriftung = tk.Label(fenster, text='Bitte klicken')
beschriftung.pack()

knopf = tk.Button(fenster, text='Klick mich', command=klick)
knopf.pack()

fenster.mainloop()
```

Der Text in einem Widget kann auch unser Speicher sein. Dieser Zähler liest die Zahl aus der Beschriftung, erhöht sie und schreibt sie zurück.

```python
import tkinter as tk

def plus():
    zahl = int(anzeige['text'])
    anzeige['text'] = zahl + 1

fenster = tk.Tk()
fenster.title('Zähler')
fenster.geometry('300x120')

anzeige = tk.Label(fenster, text='0')
anzeige.pack()

knopf = tk.Button(fenster, text='+1', command=plus)
knopf.pack()

fenster.mainloop()
```

### 3. Text eingeben

Ein `Entry` ist ein Eingabefeld. Mit `eingabe.get()` holen wir den eingegebenen Text. Wie bei `input()` ist das Ergebnis immer ein **String**.

```python
import tkinter as tk

def begruessen():
    name = eingabe.get()
    ausgabe['text'] = 'Hallo ' + name

fenster = tk.Tk()
fenster.title('Begrüßung')
fenster.geometry('300x120')

frage = tk.Label(fenster, text='Wie heißt du?')
frage.pack()

eingabe = tk.Entry(fenster)
eingabe.pack()

knopf = tk.Button(fenster, text='Los', command=begruessen)
knopf.pack()

ausgabe = tk.Label(fenster, text='')
ausgabe.pack()

fenster.mainloop()
```

### 4. Mit Zahlen rechnen

Zum Rechnen wandeln wir den Text mit `int()` oder `float()` um. Der Aufbau ist bei solchen Programmen immer gleich: Eingabe holen, verarbeiten, Ausgabe anzeigen.

```python
import tkinter as tk

def rechnen():
    euro = float(eingabe.get())
    dollar = euro * 1.1
    ausgabe['text'] = f'{dollar:.2f} Dollar'

fenster = tk.Tk()
fenster.title('Umrechner')
fenster.geometry('300x120')

eingabe = tk.Entry(fenster)
eingabe.pack()

knopf = tk.Button(fenster, text='Umrechnen', command=rechnen)
knopf.pack()

ausgabe = tk.Label(fenster, text='')
ausgabe.pack()

fenster.mainloop()
```

### 5. Beispiel: Zahlenraten

Wir verbinden das Gelernte zu einem kleinen Spiel. Das Programm denkt sich eine Zufallszahl, wir raten, und es antwortet mit „zu klein“, „zu groß“ oder „richtig“.

```python
import random
import tkinter as tk

zahl = random.randint(1, 100)

def pruefen():
    tipp = int(eingabe.get())
    if tipp < zahl:
        ausgabe['text'] = 'zu klein'
    elif tipp > zahl:
        ausgabe['text'] = 'zu groß'
    else:
        ausgabe['text'] = 'richtig'

fenster = tk.Tk()
fenster.title('Zahlenraten')
fenster.geometry('300x120')

frage = tk.Label(fenster, text='Zahl von 1 bis 100:')
frage.pack()

eingabe = tk.Entry(fenster)
eingabe.pack()

knopf = tk.Button(fenster, text='Prüfen', command=pruefen)
knopf.pack()

ausgabe = tk.Label(fenster, text='')
ausgabe.pack()

fenster.mainloop()
```

#### Zusammenfassung

| | |
|---|---|
| GUI | grafische Benutzeroberfläche mit Fenstern und Widgets |
| `tk.Tk()` | erzeugt das Fenster |
| `tk.Label(fenster, text=...)` | Beschriftung |
| `tk.Button(fenster, text=..., command=f)` | Knopf, ruft beim Klicken die Funktion `f` auf |
| `tk.Entry(fenster)` | Eingabefeld, `eingabe.get()` liefert den Text als String |
| `widget.pack()` | setzt das Widget ins Fenster |
| `widget['text'] = ...` | ändert den angezeigten Text |
| `fenster.mainloop()` | zeigt das Fenster und wartet auf Aktionen, steht am Ende |
